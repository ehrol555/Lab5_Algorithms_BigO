```python
import random
import time

# --- Prerequisites (Ensure these match your student inputs) ---
LAST_NAME = "TABARDILLA"
STUDENT_ID = "TUPM-26-7938"
FAVORITE_ARTIST = "ARTHUR NERY"

STUDENT_ID_DIGITS = "".join(ch for ch in STUDENT_ID if ch.isdigit())
SEED_DIGIT = int(STUDENT_ID_DIGITS[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID_DIGITS)
NAME_LENGTH = len(LAST_NAME.strip())
ARTIST_LENGTH = len(FAVORITE_ARTIST.strip())

# --- Data Generation Rules ---
base_orders = [
    ID_SUM + i * (ARTIST_LENGTH + 1)
    for i in range(1, 11)
]

random.seed(SEED_DIGIT)
datasets = {
    "Already Sorted": sorted(base_orders),
    "Reverse Sorted": sorted(base_orders, reverse=True),
    "Random Order": random.sample(base_orders, len(base_orders)),
}

# --- Sorting Algorithms ---
def bubble_sort(data):
    arr = data.copy()
    comparisons = swaps = 0
    for i in range(len(arr)):
        for j in range(len(arr) - 1):
            comparisons += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1
    return arr, comparisons, swaps

def selection_sort(data):
    arr = data.copy()
    comparisons = swaps = 0
    for i in range(len(arr) - 1):
        minimum = i
        for j in range(i + 1, len(arr)):
            comparisons += 1
            if arr[j] < arr[minimum]:
                minimum = j
        if minimum != i:
            arr[i], arr[minimum] = arr[minimum], arr[i]
            swaps += 1
    return arr, comparisons, swaps

def insertion_sort(data):
    arr = data.copy()
    comparisons = shifts = 0
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                shifts += 1
                j -= 1
            else:
                break
        arr[j + 1] = key
    return arr, comparisons, shifts

algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort,
}

# --- Assessment Data Execution & Formatting ---
print("=== ASSESSMENT DATA: SORTING ALGORITHMS ==-\n")
print(f"Generated Base Orders: {base_orders}\n")

for data_name, data_vals in datasets.items():
    print(f"--- Input Condition: {data_name} ---")
    print(f"Raw Input: {data_vals}")
    
    for algo_name, algo_func in algorithms.items():
        start_time = time.time()
        sorted_arr, comparisons, movements = algo_func(data_vals)
        exec_time = time.time() - start_time
        
        is_sorted = (sorted_arr == sorted(data_vals))
        
        print(f"  * Algorithm: {algo_name}")
        print(f"    - Sorted Output: {sorted_arr}")
        print(f"    - Comparison Counts: {comparisons}")
        print(f"    - Movement Counts (Swaps/Shifts): {movements}")
        print(f"    - Execution Time: {exec_time:.6f} seconds")
        print(f"    - Sorting Verification: {is_sorted}")
    print()

print("=== BIG O ANALYSIS & PERFORMANCE SUMMARY ===")
print("Bubble Sort    -> Time: O(n^2) | Space: O(1)")
print("Selection Sort -> Time: O(n^2) | Space: O(1)")
print("Insertion Sort -> Time: O(n^2) worst/average, O(n) best | Space: O(1)")

```
